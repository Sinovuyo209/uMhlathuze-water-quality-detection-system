```python
import pandas as pd
import numpy as np
import os


excel_files = [f for f in os.listdir('.') if f.endswith('.xlsx') or f.endswith('.xls')]

if excel_files:
    file_to_use = excel_files[0]
    xls = pd.ExcelFile(file_to_use)
    df = pd.read_excel(file_to_use, sheet_name=xls.sheet_names[0])
    df.columns = df.columns.str.strip()
    print(f"Successfully loaded dataset from: {file_to_use}")

    # Calculate Interquartile Range (IQR) for outlier detection
    numeric_cols = ['EC', 'TDS', 'pH', 'NO3', 'Al']
    outlier_summary = {}

    for col in numeric_cols:
        if col in df.columns:
            Q1 = df[col].quantile(0.25)
            Q3 = df[col].quantile(0.75)
            IQR = Q3 - Q1
            lower_bound = Q1 - 1.5 * IQR
            upper_bound = Q3 + 1.5 * IQR
            outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
            outlier_summary[col] = len(outliers)

    print("\nOutlier counts per key parameter (IQR method):")
    for k, v in outlier_summary.items():
        print(f"- {k}: {v} potential outlier records")
else:
    print("Error: No Excel file found in the current directory.")
```

```python
from sklearn.preprocessing import MinMaxScaler

X = df.drop('Y', axis=1)
y = df['Y']

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

print("Input shape:", X_scaled.shape)
print("Target shape:", y.shape)
print("Scaling done!")
```

```python
import subprocess
subprocess.call(['pip', 'install', 'imbalanced-learn'])
```

```python
from imblearn.over_sampling import SMOTE
import pandas as pd

X_temp = pd.DataFrame(X_scaled)
y_temp = y.reset_index(drop=True)

X_temp = X_temp[y_temp != 3]
y_temp = y_temp[y_temp != 3]

target_samples = 300
sampling_dict = {0: target_samples, 1: target_samples, 2: target_samples}

smote = SMOTE(sampling_strategy=sampling_dict, random_state=42)
X_resampled, y_resampled = smote.fit_resample(X_temp, y_temp)

print("Before SMOTE:", y_temp.value_counts().sort_index().to_dict())
print("After SMOTE:", y_resampled.value_counts().sort_index().to_dict())
print("Total synthetic samples after SMOTE expansion:", len(X_resampled))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_resampled, y_resampled, test_size=0.2, random_state=42)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier

rf = RandomForestClassifier(random_state=42)
dt = DecisionTreeClassifier(random_state=42)
xgb = XGBClassifier(random_state=42, eval_metric='mlogloss')

rf.fit(X_train, y_train)
dt.fit(X_train, y_train)
xgb.fit(X_train, y_train)

print("All 3 models trained successfully!")
```

```python
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score

models = {'Random Forest': rf, 'Decision Tree': dt, 'XGBoost': xgb}

for name, model in models.items():
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred, average='weighted')
    precision = precision_score(y_test, y_pred, average='weighted')
    recall = recall_score(y_test, y_pred, average='weighted')
    
    print(f"{name}")
    print(f"  Accuracy  : {acc:.2f}")
    print(f"  F1 Score  : {f1:.2f}")
    print(f"  Precision : {precision:.2f}")
    print(f"  Recall    : {recall:.2f}")
    print()
```

```python
from sklearn.model_selection import cross_val_score
import numpy as np

models = {'Random Forest': rf, 'Decision Tree': dt, 'XGBoost': xgb}

for name, model in models.items():
    scores = cross_val_score(model, X_resampled, y_resampled, cv=5, scoring='accuracy')
    print(f"{name}")
    print(f"  Fold Scores : {scores.round(2)}")
    print(f"  Mean Accuracy : {scores.mean():.2f}")
    print(f"  Std : {scores.std():.2f}")
    print()
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
model_names = ['Random Forest', 'Decision Tree', 'XGBoost']
models_list = [rf, dt, xgb]

for i, (name, model) in enumerate(zip(model_names, models_list)):
    y_pred = model.predict(X_test)
    cm = confusion_matrix(y_test, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', ax=axes[i], cmap='Blues')
    axes[i].set_title(name)
    axes[i].set_xlabel('Predicted')
    axes[i].set_ylabel('Actual')

plt.tight_layout()
plt.savefig('confusion_matrices.png')
plt.show()
print("Done!")
```

```python
import matplotlib.pyplot as plt
import numpy as np

models = {'Decision Tree': dt, 'Random Forest': rf, 'XGBoost': xgb}
features = X.columns

fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharey=True)

for i, (name, model) in enumerate(models.items()):
    importances = model.feature_importances_
    sorted_idx = np.argsort(importances)[::-1]
    axes[i].bar(range(X.shape[1]), importances[sorted_idx], color='#2b5c8f')
    axes[i].set_title(name, fontsize=12, fontweight='bold')
    axes[i].set_xticks(range(X.shape[1]))
    axes[i].set_xticklabels(features[sorted_idx], rotation=90, fontsize=9)
    axes[i].grid(axis='y', linestyle='--', alpha=0.6)

axes[0].set_ylabel("Importance Score", fontsize=11)
plt.tight_layout()
plt.savefig('all_models_feature_importance.png', dpi=300)
plt.show()
```
