```python
import pandas as pd

df = pd.read_excel('water_quality_dataset.xlsx')

print("Number of rows and columns:", df.shape)
print("\nColumn names:")
print(list(df.columns))
print("\nFirst 5 rows of data:")
df.head()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nWater Quality Classes (Y column):")
print(df['Y'].value_counts().sort_index())

print("\nWhat each class means:")
print("Class 0 = Excellent water quality")
print("Class 1 = Good water quality")
print("Class 2 = Poor water quality")
print("Class 3 = Very poor water quality")
```

```python
import matplotlib.pyplot as plt

class_counts = df['Y'].value_counts().sort_index()
labels = ['Class 0\nExcellent', 'Class 1\nGood', 'Class 2\nPoor', 'Class 3\nVery Poor']
colors = ['green', 'blue', 'orange', 'red']

plt.figure(figsize=(8, 5))
plt.bar(labels, class_counts.values, color=colors)
plt.title('Water Quality Class Distribution - uMhlathuze River Dataset')
plt.xlabel('Water Quality Class')
plt.ylabel('Number of Samples')
plt.tight_layout()
plt.savefig('class_distribution.png')
plt.show()

print("Chart saved!")
```

```python
stats = df.describe().round(2)
print(stats)
```

```python
import seaborn as sns

plt.figure(figsize=(12, 8))
sns.heatmap(df.corr().round(2), annot=True, cmap='coolwarm', fmt='.1f', linewidths=0.5)
plt.title('Correlation Between Water Quality Parameters')
plt.tight_layout()
plt.savefig('correlation_heatmap.png')
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler

X = df.drop('Y', axis=1)
y = df['Y']

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

print("Input shape:", X_scaled.shape)
print("Target shape:", y.shape)
print("Scaling done!")
```

```python
import subprocess
subprocess.call(['pip', 'install', 'imbalanced-learn'])
```

```python
from imblearn.over_sampling import SMOTE
import pandas as pd

X_temp = pd.DataFrame(X_scaled)
y_temp = y.reset_index(drop=True)

X_temp = X_temp[y_temp != 3]
y_temp = y_temp[y_temp != 3]

smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X_temp, y_temp)

print("Before SMOTE:", y_temp.value_counts().sort_index().to_dict())
print("After SMOTE:", y_resampled.value_counts().sort_index().to_dict())
print("Total samples after SMOTE:", len(X_resampled))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_resampled, y_resampled, test_size=0.2, random_state=42)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier

rf = RandomForestClassifier(random_state=42)
dt = DecisionTreeClassifier(random_state=42)
xgb = XGBClassifier(random_state=42, eval_metric='mlogloss')

rf.fit(X_train, y_train)
dt.fit(X_train, y_train)
xgb.fit(X_train, y_train)

print("All 3 models trained successfully!")
```

```python
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score

models = {'Random Forest': rf, 'Decision Tree': dt, 'XGBoost': xgb}

for name, model in models.items():
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred, average='weighted')
    precision = precision_score(y_test, y_pred, average='weighted')
    recall = recall_score(y_test, y_pred, average='weighted')
    
    print(f"{name}")
    print(f"  Accuracy  : {acc:.2f}")
    print(f"  F1 Score  : {f1:.2f}")
    print(f"  Precision : {precision:.2f}")
    print(f"  Recall    : {recall:.2f}")
    print()
```

```python
from sklearn.model_selection import cross_val_score
import numpy as np

models = {'Random Forest': rf, 'Decision Tree': dt, 'XGBoost': xgb}

for name, model in models.items():
    scores = cross_val_score(model, X_resampled, y_resampled, cv=5, scoring='accuracy')
    print(f"{name}")
    print(f"  Fold Scores : {scores.round(2)}")
    print(f"  Mean Accuracy : {scores.mean():.2f}")
    print(f"  Std : {scores.std():.2f}")
    print()
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
model_names = ['Random Forest', 'Decision Tree', 'XGBoost']
models_list = [rf, dt, xgb]

for i, (name, model) in enumerate(zip(model_names, models_list)):
    y_pred = model.predict(X_test)
    cm = confusion_matrix(y_test, y_pred)
    sns.heatmap(cm, annot=True, fmt='d', ax=axes[i], cmap='Blues')
    axes[i].set_title(name)
    axes[i].set_xlabel('Predicted')
    axes[i].set_ylabel('Actual')

plt.tight_layout()
plt.savefig('confusion_matrices.png')
plt.show()
print("Done!")
```

```python
import pandas as pd

feature_names = df.drop('Y', axis=1).columns

rf_importance = pd.Series(rf.feature_importances_, index=feature_names).sort_values(ascending=True)

plt.figure(figsize=(8, 8))
rf_importance.plot(kind='barh', color='steelblue')
plt.title('Feature Importance - Random Forest')
plt.xlabel('Importance Score')
plt.tight_layout()
plt.savefig('feature_importance.png')
plt.show()
print("Done!")
```

```python
models_names = ['Random Forest', 'Decision Tree', 'XGBoost']
accuracies = [0.92, 0.82, 0.96]
f1_scores = [0.92, 0.82, 0.96]

x = range(len(models_names))
width = 0.35

plt.figure(figsize=(9, 5))
plt.bar([i - width/2 for i in x], accuracies, width, label='Accuracy', color='steelblue')
plt.bar([i + width/2 for i in x], f1_scores, width, label='F1 Score', color='orange')
plt.xticks(x, models_names)
plt.ylim(0, 1.1)
plt.ylabel('Score')
plt.title('Model Comparison - Accuracy and F1 Score')
plt.legend()
plt.tight_layout()
plt.savefig('model_comparison.png')
plt.show()
print("Done!")
```
